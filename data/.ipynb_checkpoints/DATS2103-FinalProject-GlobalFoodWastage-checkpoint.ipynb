```python
import pandas as pd
import numpy as np 
import math
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split, KFold

from sklearn.linear_model import LinearRegression, Ridge
from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import mean_squared_error, r2_score
import statsmodels.api as sm
```

# Data Pre-Processing
Data downloaded from: https://www.kaggle.com/datasets/atharvasoundankar/global-food-wastage-dataset-2018-2024

```python
rawdf = pd.read_csv("GlobalFoodWastage.csv")
rawdf
```

## One-Hot Encoded Variables

```python
datadf = rawdf[["Country", "Year", "Food Category", "Total Waste (Tons)", "Economic Loss (Million $)", "Avg Waste per Capita (Kg)", "Population (Million)", "Household Waste (%)"]]

# hot encoded vars
datadf = pd.get_dummies(datadf, columns=["Country", "Food Category"], prefix=["Country", "Food Category"], dtype=int)
datadf
```

## Standardization

```python
X = datadf.drop(columns=["Economic Loss (Million $)"])
y = datadf[["Economic Loss (Million $)"]]

predictors = X.columns.tolist()
print("Number of Predictors:", len(predictors))

standardized_X = (X - X.mean()) / X.std()

for p in predictors:
    
    datadf[p] = standardized_X[p]
print("Std Dev for X_0 standardization:", datadf["Total Waste (Tons)"].std())

X
```

# K-Fold Cross Validation

```python
kf_object = KFold(n_splits=5)
fwdata = {}

for k, (train_index, test_index) in enumerate(kf_object.split(datadf)):

    currtrain = datadf.iloc[train_index]
    currtest = datadf.iloc[test_index]

    currtrainX = currtrain.drop(columns=["Economic Loss (Million $)"])
    currtestX = currtest.drop(columns=["Economic Loss (Million $)"])

    currtrainy = currtrain[["Economic Loss (Million $)"]]
    currtesty = currtest[["Economic Loss (Million $)"]]

    fwdata["X_train_split" + str(k)] = currtrainX
    fwdata["X_test_split" + str(k)] = currtestX

    fwdata["y_train_split" + str(k)] = currtrainy
    fwdata["y_test_split" + str(k)] = currtesty
```

# Model #1 - Linear Regression

```python
LR_MSEs = []
LR_R2s = []

for k in range(0, 5):

    currtrainX = fwdata["X_train_split" + str(k)]
    currtrainy = fwdata["y_train_split" + str(k)]
    currtestX = fwdata["X_test_split" + str(k)]
    currtesty = fwdata["y_test_split" + str(k)]

    LR_object = LinearRegression().fit(currtrainX,  currtrainy["Economic Loss (Million $)"].tolist())

    y_pred_LR = LR_object.predict(currtestX)
    currMSE = mean_squared_error(currtesty, y_pred_LR)
    currR2 = r2_score(currtesty, y_pred_LR)

    LR_MSEs.append(currMSE)
    LR_R2s.append(currR2)

    print("=== Current Fold:", k,"===")
    print("MSE Score: $", currMSE)
    print("R2:", currR2)

print("=====")
print("Average MSE Score: $", sum(LR_MSEs)/5)
print("Average  R2:", sum(LR_R2s)/5)
bestMSE = min(LR_MSEs)
bestMSE_index = LR_MSEs.index(bestMSE)

print("Fold with Lowest MSE: k =", bestMSE_index, ", MSE:", bestMSE)
```

# Model #2 - Ridge Regression

```python
RR_MSEs = []
RR_R2s = []

for k in range(0, 5):

    currtrainX = fwdata["X_train_split" + str(k)]
    currtrainy = fwdata["y_train_split" + str(k)]
    currtestX = fwdata["X_test_split" + str(k)]
    currtesty = fwdata["y_test_split" + str(k)]

    RR_object = Ridge(alpha=1).fit(currtrainX, currtrainy["Economic Loss (Million $)"].tolist())

    y_pred_RR = RR_object.predict(currtestX)
    currMSE = mean_squared_error(currtesty, y_pred_RR)
    currR2 = r2_score(currtesty, y_pred_RR)

    RR_MSEs.append(currMSE)
    RR_R2s.append(currR2)

    print("=== Current Fold:", k,"===")
    print("MSE Score: $", currMSE)
    print("R2:", currR2)

print("=====")
print("Average MSE Score: $", sum(RR_MSEs)/5)
print("Average  R2:", sum(RR_R2s)/5)
bestMSE = min(RR_MSEs)
bestMSE_index = RR_MSEs.index(bestMSE)

print("Fold with Lowest MSE: k =", bestMSE_index, ", MSE:", bestMSE)
```

# Model #3 - Random Forest

```python
RF_MSEs = []
RF_R2s = []
for k in range(0, 5):

    currtrainX = fwdata["X_train_split" + str(k)]
    currtrainy = fwdata["y_train_split" + str(k)]
    currtestX = fwdata["X_test_split" + str(k)]
    currtesty = fwdata["y_test_split" + str(k)]
    
    RF_object = RandomForestRegressor(n_estimators=100, random_state=42).fit(currtrainX,  currtrainy["Economic Loss (Million $)"].tolist())
    y_pred = RF_object.predict(currtestX)

    currMSE = mean_squared_error(currtesty, y_pred)
    currR2 = r2_score(currtesty, y_pred)

    RF_MSEs.append(currMSE)
    RF_R2s.append(currR2)
    print("=== Current Fold:", k,"===")
    print("MSE Score: $", currMSE)
    print("R2: ", currR2)

print("Average MSE across 5 folds for Random Forest: $", sum(RF_MSEs)/5)
print("Average R2 across 5 folds for Random Forest: $", sum(RF_R2s)/5)
bestMSE = min(RF_MSEs)
bestMSE_index = RF_MSEs.index(bestMSE)

print("Fold with Lowest MSE: k =", bestMSE_index, ", MSE:", bestMSE)
```
