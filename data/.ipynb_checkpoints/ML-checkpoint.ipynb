Let's see if we can have the computer learn digit recognition.

We'll use something called a *support vector classifier (SVC)*.

```python
from sklearn.svm import SVC

# svm stands for "support vector machine"
# SVC stands for "support vector classifier"
# clf stands for "classifier"

clf = SVC(kernel = 'linear')
print(type(clf))
```

This has instantiated an instance of the class `sklearn.svm.classes.SVC`.

In scikit-learn, 
an *estimator for classification* 
is a Python object that implements the methods 
`fit(X_train, y)` and `predict(X_test)`.

Suppose we want to *train* an SVC on `n_samples` samples 
and that each sample has `n_features` features. 

We would call `fit(X_train, y)`.
 - `X_train` should have shape `(n_samples, n_features)`.
 - `y` should have shape `(n_samples,)`.
 - `fit` will return `self`, the instance of the estimator.
   After the training, it will have a more sensible `predict` method.

```python
imgs = digits.images            # not relevant for training and testing; we need the flattened version (data).
data = digits.data
targ = digits.target
```

```python
clf.fit(data[:-2], targ[:-2])   # train using all but the last 2 samples.
```

Suppose we now want to *test* how accurate our SVC is on 
`n_samples` samples (potentially a different number than above) 

and that each sample has `n_features` features. 
We would call `predict(X_test)`.
 - `X_test` should have shape `(n_samples, n_features)`.
 - `predict` returns something of shape `(n_samples,)`,
   what it thinks are the correct targets.

Above, we trained on all but the last `2` samples.

It is interesting to wonder how much training data the SVC needs to see before it does a good job.

(I've reloaded all modules so this cell can be run independently of everything above.)

```python
import numpy as np
import matplotlib.pyplot as plt


from sklearn import datasets
from sklearn.svm import SVC


digs = datasets.load_digits()
data = digs.data
targ = digs.target


def accuracy_after_train_on_first_n_test_on_rest(n):
    clf = SVC(kernel = 'linear')
    
    clf.fit(data[:n], targ[:n])
    
    predict_targs = clf.predict(data[n:])
    correct_targs = targ[n:]
    
    return 100 * np.count_nonzero(predict_targs == correct_targs) / len(correct_targs)


n_vals = range(10, len(data), 10)
accuracy_vals = [accuracy_after_train_on_first_n_test_on_rest(n) for n in n_vals]

plt.plot(n_vals, accuracy_vals)
```

We see that it had `90%` accuracy after training on `400` digits.

## SVC: an artificial example where it does a good job

Without understanding (to some extent) what a function does, 
I won't find using that function particularly satisfying.

Above, we used `clf = SVC(kernel = 'linear')`, `clf.fit`, and `clf.predict`.

The goal of this section is to give you some idea of what these functions are doing,

but not in so much detail that we're in a math class.

Let's generate some of our own data in the plane: 
data that can be separated by a line.

```python
import numpy as np

n = 2000 # number of samples to train and test on.

upperRight_train = np.random.random((n//2, 2))
lowerLeft_train = -np.random.random((n//2, 2))

# remember: the first axis enumerates the samples;
# the second axis enumerates the features of the data.

# so upperRight_train has n/2 samples with 2 features each.

# np.random.random (https://docs.scipy.org/doc/numpy-1.16.0/reference/generated/numpy.random.random.html)
# returns arrays consisting of random floats in the interval 0.0 <= x < 1.0.

# So we have n/2 samples in the upper right quadrant
# and n/2 samples in the lower left quadrant.

upperRight_labels = np.ones((n//2,), dtype = np.int8)
lowerLeft_labels = -np.ones((n//2,), dtype = np.int8)
```

It'd be nice if we could see our data, so let's use scatter plot.

```python
import matplotlib.pyplot as plt

# https://matplotlib.org/3.2.1/api/_as_gen/matplotlib.pyplot.scatter.html
# https://matplotlib.org/3.2.1/api/_as_gen/matplotlib.pyplot.axis.html

# https://docs.python.org/3.7/tutorial/controlflow.html#arbitrary-argument-lists

def draw_data(*d, xmin = -1, xmax = 1, ymin = -1, ymax = 1):
    plt.axis([xmin, xmax, ymin, ymax])

    for data in d:
        plt.scatter(data[:,0], data[:,1], s=1)
```

```python
draw_data(upperRight_train, lowerLeft_train)
```

Let's assemble our training set and targets.

```python
# https://numpy.org/doc/stable/reference/generated/numpy.concatenate.html

train = np.concatenate((upperRight_train, lowerLeft_train))
targ = np.concatenate((upperRight_labels, lowerLeft_labels))
```

Imagine I was training you to understand the difference between cats and dogs. 

Should I give you `1000` cats followed by `1000` dogs, 
or should I give you a random bunch of cats and and dogs labelled correctly?

It is unclear to me which is better for the SVM algorithm. 
I suspect it does not matter much. 
But, out of paranoia and as an excuse to use `np.random.permutation` 
and some fancy indexing, let's randomize the order of our samples.

```python
from sklearn.svm import SVC
clf = SVC(kernel = 'linear')

# https://docs.scipy.org/doc/numpy-1.16.0/reference/generated/numpy.random.permutation.html
# we use fancy indexing to shuffle the data and targets in the same way.

indices = np.random.permutation(n)
clf.fit(train[indices], targ[indices])
```

Let's test to see how the SVC did.

We generate a testing set, and see what the SVC predicts for its targets.

```python
upperRight_test = np.random.random((n//2, 2))
lowerLeft_test = -np.random.random((n//2, 2))

test = np.concatenate((upperRight_test, lowerLeft_test))

print('train == test is', np.all(train == test)) # checking our test data is different from our training data!!

print('accuracy ==', 100 * np.count_nonzero( clf.predict(test) == targ ) / n)
```

100%! How did it do this?



`clf`, an instance of `sklearn.svm.classes.SVC`, has many attributes. 
Below I've printed some of them, as well as various relations between them. 

Together with `SVMs.pdf`, this is enough to explain what's going to the mathematically inclined student. 

However, don't worry if you're not a mathematician...

I'll draw a picture of the most important information immediately after this, 
namely `clf.coeff_` and `clf.intercept_` (which describe a line).

```python
print('(Where data is overwhelming, I have printed the shape;')
print('of course, you can look at the actual data if you like.)')
print('')

print('The target values:')
print(clf.classes_, '\n')

print('The shape of training data:')
print(clf.shape_fit_, '\n')

print('The number of samples used from each target group:')
print(clf.n_support_, '\n')

print('The indices of the samples used, the actual samples used, a check that this makes sense:')
print(clf.support_.shape)
print(clf.support_vectors_.shape)
print(np.all(train[indices[clf.support_]] == clf.support_vectors_))
print('')

print('The coefficients of the "support vectors"')
print(clf.dual_coef_.shape, '\n')

print('The coefficients of the separating line, and a demonstration of how these are deduced')
print(clf.coef_.shape)
print(np.all(np.sum(clf.dual_coef_.reshape(-1,1) * clf.support_vectors_, axis = 0) == clf.coef_.reshape(-1)))
print('')

print('The intercept of the separating line')
print(clf.intercept_.shape)
```

The equation of the *separating line* is given by:

`clf.coef_[0,0] * x + clf.coef_[0,1] * y + clf.intercept_[0] = 0.`

```python
def separating_line(clf):
    x = np.linspace(-1, 1, 2049) # (the fact that 2048 is a power of 2 avoids rounding errors)
    y = -(clf.coef_[0,0] * x  +  clf.intercept_[0]) / clf.coef_[0,1]
    return np.concatenate((x[:,np.newaxis], y[:,np.newaxis]), axis = -1)
```

```python
line = separating_line(clf)
draw_data(upperRight_train, lowerLeft_train, line)
```

`clf.fit` ran an algorithm to calculate this line. 
Hopefully, it separates the data into the two groups.

It then uses this line to make future `predict`ions.

Let's zoom in quite a lot.

```python
draw_data(upperRight_train, lowerLeft_train, line, xmin = -0.05, xmax = 0.05, ymin = -0.05, ymax = 0.05)
```

Hopefully, you can tell from the picture which group `clf` will predict `[0, 0]` belongs to.

```python
print('[0, 0] belongs to group', clf.predict([[0,0]])[0])
```

## SVC: an artificial example where it does a bad job

Refreshing the variables is okay because I've redefined the functions from above that I want to reuse.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.svm import SVC

def draw_data(*d, xmin = -1, xmax = 1, ymin = -1, ymax = 1):
    plt.axis([xmin, xmax, ymin, ymax])

    for data in d:
        plt.scatter(data[:,0], data[:,1], s=1)

def separating_line(clf):
    x = np.linspace(-1, 1, 2049)
    y = -(clf.coef_[0,0] * x  +  clf.intercept_[0]) / clf.coef_[0,1]
    return np.concatenate((x[:,np.newaxis], y[:,np.newaxis]), axis = -1)
```

We generate two classes of samples and the corresponding labels.

```python
n = 4000 # number of samples to train and test on.

inner_train = np.array([-1/5, -1/5])  +  2 * np.random.random((n//2, 2)) / 5

ot1 = np.array([-1  , -1/3])  +  2 * np.random.random((n//8, 2)) / 3
ot2 = np.array([-1/3, -1  ])  +  2 * np.random.random((n//8, 2)) / 3
ot3 = np.array([-1/3,  1/3])  +  2 * np.random.random((n//8, 2)) / 3
ot4 = np.array([ 1/3, -1/3])  +  2 * np.random.random((n//8, 2)) / 3

outer_train = np.concatenate((ot1, ot2, ot3, ot4))

inner_labels = np.ones((n//2,), dtype = np.int8)
outer_labels = -np.ones((n//2,), dtype = np.int8)
```

Those equations might look terrible to you,
but this is what they generated...

```python
draw_data(inner_train, outer_train)
```

We assemble our training set and targets. 

```python
train = np.concatenate((inner_train, outer_train))
targ = np.concatenate((inner_labels, outer_labels))
```

We train our SVD.

```python
clf = SVC(kernel = 'linear')

indices = np.random.permutation(n)
clf.fit(train[indices], targ[indices])
```

We generate a test set and test our SVD.

```python
inner_test = np.array([-1/5, -1/5])  +  2 * np.random.random((n//2, 2)) / 5

ot1 = np.array([-1  , -1/3])  +  2 * np.random.random((n//8, 2)) / 3
ot2 = np.array([-1/3, -1  ])  +  2 * np.random.random((n//8, 2)) / 3
ot3 = np.array([-1/3,  1/3])  +  2 * np.random.random((n//8, 2)) / 3
ot4 = np.array([ 1/3, -1/3])  +  2 * np.random.random((n//8, 2)) / 3

outer_test = np.concatenate((ot1, ot2, ot3, ot4))

test = np.concatenate((inner_test, outer_test))

print('train == test is', np.all(train == test)) # just checking our test data is different from our training data!!

print('accuracy ==', 100 * np.count_nonzero( clf.predict(test) == targ ) / n)
```

Why did it do so badly? A picture is worth a thousand words.

```python
line = separating_line(clf)
draw_data(inner_train, outer_train, line)
```

## Preprocessing to help SVC: an artificial example

We're going to keep using the variables from above, so don't refresh them.

Although the SVC is not successful at separating the date above, 
by *preprocessing* our data we can make it do much better.

```python
def preprocess(data):
    return data
```

Currently, our preprocessing function does nothing.

The following code trains a new SVC using the same training data as before, 
and tests it on the same data as before. 

Unsurprisingly, the result is the same.

```python
clf = SVC(kernel = 'linear')

clf.fit(preprocess(train)[indices], targ[indices])
print('accuracy ==', 100 * np.count_nonzero( clf.predict(preprocess(test)) == targ ) / n)
```

Now I apply some preprocessing. 

We replace each sample `(x, y)` (which has two features) by `(x, y, x**2 + y**2)` (which has 3 features).

Our samples now live in 3D space. 
The outer samples are higher up than then inner samples. 

So it might be possible to separate the two groups using a `plane`.

(A hyperplane in 2 dimensions is a line, 
a hyperplane in 3 dimensions is a plane, 

a hyperplane in higher dimensions feels more difficult unless you're a mathematician.

In general, the linear SVC separates data using a hyperplane.)

```python
def preprocess(data):
    z = data[:,0]**2 + data[:,1]**2
    return np.concatenate((data, z[:,np.newaxis]), axis = -1)
```

```python
clf = SVC(kernel = 'linear')

indices = np.random.permutation(n)
clf.fit(preprocess(train)[indices], targ[indices])

print('accuracy ==', 100 * np.count_nonzero( clf.predict(preprocess(test)) == targ ) / n)
```

Woah! From <70% accurate to ~99% accurate!

I cannot draw a picture on here very easily, but I will try and draw a bad picture on the whiteboard.

Of course, here, I made my own data, and I made it exactly so that my preprocessing trick would work.

In real life, you may have some medical data...

 - If you don't understand the data very well, preprocessing might be reduced to trying things almost at random. 

 - If you understand your data a little better, you might try some preprocessing that you think will help.

   Hopefully it does and you can cure a deadly disease.

We can preprocess even better by replacing each sample `(x, y)` by `(x, y, sqrt(x**2 + y**2))`.

```python
def preprocess(data):
    z = np.sqrt(data[:,0]**2 + data[:,1]**2)
    return np.concatenate((data, z[:,np.newaxis]), axis = -1)
```

```python
clf = SVC(kernel = 'linear')

indices = np.random.permutation(n)
clf.fit(preprocess(train)[indices], targ[indices])

print('accuracy ==', 100 * np.count_nonzero( clf.predict(preprocess(test)) == targ ) / n)
```

This does better because the spread of the `z`-coordinate values is less clustered around `0` 
and more evenly spread between the values `0` and `1`.
