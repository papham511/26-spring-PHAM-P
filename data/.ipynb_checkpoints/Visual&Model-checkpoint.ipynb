# Visualize Plot 

## 1. Introduction & Data Preparation

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
users_df = pd.read_csv("users_cleaned_dataset.csv")
fusers_df = pd.read_csv("fusers_cleaned_dataset.csv")

# Combined data
combined_df = pd.concat([users_df, fusers_df])
combined_df
```

## 2. Visualization

```python
# 2. Scatter Plot
followers = combined_df['followers_count'] + 1     # adding 1 to avoid log(0) errors
following = combined_df['following_count'] + 1
labels = combined_df['fake']

plt.scatter(following[labels == 0], followers[labels == 0], 
            color='blue', alpha=0.5, label='Real', s=10)
plt.scatter(following[labels == 1], followers[labels == 1], 
            color='red', alpha=0.5, label='Fake', s=10)

# Apply log scales
plt.xscale('log')
plt.yscale('log')

# Label
plt.xlabel('Following Count (Log Scale)')
plt.ylabel('Followers Count (Log Scale)')
plt.title('Followers vs. Following')
plt.legend()

plt.show()
```

```python
# 4. Histogram
users = pd.read_csv("users_cleaned_dataset.csv")
fusers = pd.read_csv("fusers_cleaned_dataset.csv")

fusers['User Type'] = 'Fake'
users['User Type'] = 'Real'

combined_df = pd.concat([users, fusers], ignore_index=True)

df_filtered = combined_df[combined_df['num_posts'] >= 0].copy()      #Filter out unknown (-1)

df_filtered['num_posts_log_adj'] = df_filtered['num_posts'] + 1

plt.figure(figsize=(12, 6))
sns.histplot(
    data=df_filtered,
    x='num_posts_log_adj',
    hue = 'User Type',
    bins=50,
    kde=True,
    palette={'Real':'teal', 'Fake':'gold'},
    log_scale=True
)

plt.title('Distribution of Number of Posts', fontsize=14)
plt.xlabel('Number of Posts (Log Scale, +1 adjusted)', fontsize=12)
plt.ylabel('Frequency (Count)', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.3)

plt.show()
```

- Fake account (Yellow): Sharp peak at a low post count (10-100 posts) -> often newer or less active.
- Real account (teal): follow a broader, bell-shaped distribution centered much higher (1,000 - 100,000 posts) -> reflect sustain and long term user activity

```python
# 5. Heatmap
features = ['bio_length', 
            'num_posts', 
            'followers_count', 
            'following_count', 
            'username_length', 
            'num_digits_username', 
            'digit_ratio_username', 
            'followers_following_ratio', 
            'fake'
           ]

corr_df = combined_df[features].replace(-1, np.nan)   #Replace -1 (unknown) with NaN so they don't skew the correlation
corr_matrix = corr_df.corr()      #Calculate Peason correlation matrix

plt.figure(figsize=(12, 10))
sns.heatmap(
    corr_matrix, 
    annot=True,
    fmt=".2f",
    cmap='Blues',
    center=0,
    linewidths=1,
    cbar_kws={"shrink": .8}
)

plt.title('Strategic Road (Feature Correlation)', fontsize=16, pad=20)
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()

plt.show()
```

# Models

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
#Load and combine datasets

users = pd.read_csv('users_cleaned_dataset.csv')
fusers = pd.read_csv('fusers_cleaned_dataset.csv')
df = pd.concat([users, fusers])
df
```

```python
# Feature Selection
features = [
    'bio_length', 'num_posts', 'followers_count', 'following_count', 
    'username_length', 'num_digits_username', 'digit_ratio_username', 
    'followers_following_ratio'
]
X = df[features]
y = df['fake']
```

```python
# handle missing values & scaling
X = X.fillna(0)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# Train/Test Split
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)
```

```python
# Logistic Regression
lr_model = LogisticRegression()
lr_model.fit(X_train, y_train)
lr_preds = lr_model.predict(X_test)

print("Logistic Regression Accuracy:", accuracy_score(y_test, lr_preds))
print(classification_report(y_test, lr_preds))
```

```python
# Support Vector Classifier (SVC)
svc_model = SVC(kernel='rbf', probability=True)
svc_model.fit(X_train, y_train)
svc_preds = svc_model.predict(X_test)

print("SVC Accuracy:", accuracy_score(y_test, svc_preds))
print(classification_report(y_test, svc_preds))
```

```python
#Random Forest
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(X_train, y_train)
rf_preds = rf_model.predict(X_test)

print("Random Forest Accuracy:", accuracy_score(y_test, rf_preds))
print(classification_report(y_test, rf_preds))
```

```python
# Accuracy Comparison Of 3 Models
plt.clf()
final_results = {"LR":0.8736389684813753, "SVC":0.867621776504298, "Random Forest":0.9879656160458453}
sns.barplot(x=list(final_results.keys()), y=list(final_results.values()), palette='magma')
plt.title('Final Model Accuracy Comparison')
plt.ylabel('Accuracy')
plt.ylim(0.8,1.0)
plt.tight_layout()
```

```python
# Random Forest Plot
plt.clf()
sns.heatmap(confusion_matrix(y_test, rf_preds), annot=True, fmt='d', cmap='Purples')
plt.title('Random Forest')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.tight_layout()
```

```python
# Features that important for Random Forest
plt.clf()
importances = rf_model.feature_importances_
indices = np.argsort(importances)
plt.title('Random Forest: Feature Importances')
plt.barh(range(len(indices)), importances[indices])
plt.yticks(range(len(indices)), [features[i] for i in indices])
plt.xlabel('Relative Importance Score')
plt.tight_layout()
```

```python
#1. Melt the dataframe using the existing 'fake' column
bp = df.melt(id_vars= 'fake',
                      value_vars=['followers_count', 'following_count'],
                      var_name='Metric',
                      value_name='Count')

# 2. Clear current plot to ensure a fresh start 
plt.clf()
plt.figure(figsize=(10, 6))

# 3. Plot the data
plt.figure(figsize=(10, 6))
sns.boxplot(data=bp, x='fake', y='Count', hue='Metric', palette='magma')

# 4. Rename the 0 and 1 on X-axis
plt.xticks([0, 1], ['Real Accounts', 'Fake Accounts'])

plt.yscale('log')
plt.title('Comparison of Followers vs Following Counts')
plt.ylabel('Count (Log Scale)')
plt.xlabel('Account Type')
plt.tight_layout()
```
