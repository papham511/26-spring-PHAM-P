```python
import pandas as pd
import numpy as np
```

```python
# Steps:
# 1. load dataset (udf, udf1, udf2, udf3)
# 2. Rename dataset column
# 3. Merge/combine dataset into one
# 4. Clean data (NA, duplicate, convert, std column name, etc.)
# 5. Create clean data csv
# 5. Repeat for fake user dataset (fudf, fudf1, fudf2)
```

# Real Users Datasets

## Loading Datasets

```python
# DF
# Load the dataset
rawudf = pd.read_csv("users.csv")

# Choose columns need
udf = rawudf[["name", "screen_name", "statuses_count", "followers_count", "friends_count", "created_at", "description"]]

# Replace column names
udf.rename({'screen_name':'username', 'statuses_count':'num_posts', 'friends_count':'following_count', 'created_at':'account_created_at', 'description':'bio'}, axis=1, inplace=True)
udf
```

```python
# DF1
# Load the dataset
rawudf1 = pd.read_csv("users(1).csv")

# Choose columns need
udf1 = rawudf1[["name", "screen_name", "statuses_count", "followers_count", "friends_count", "created_at", "description"]]

# Replace column names
udf1.rename({'screen_name':'username', 'statuses_count':'num_posts', 'friends_count':'following_count', 'created_at':'account_created_at', 'description':'bio'}, axis=1, inplace=True)
udf1
```

```python
# DF2
# Load the dataset
rawudf2 = pd.read_csv("user(2).csv", engine="python", sep=",") #find a source for this code

# Choose columns need
udf2 = rawudf2[["name", "screen_name", "statuses_count", "followers_count", "friends_count", "created_at", "description"]]

# Replace column names
udf2.rename({'screen_name':'username', 'statuses_count':'num_posts', 'friends_count':'following_count', 'created_at':'account_created_at', 'description':'bio'}, axis=1, inplace=True)
udf2
```

```python
# DF 3
# Load the dataset
rawudf3 = pd.read_csv("insta_influencers.csv")

# Choose columns need
udf3 = rawudf3[["Account", "Username",  "Posts", "Followers", "Following"]]

# Replace column names
udf3.rename({'Account':'name', 'Username':'username', 'Posts':'num_posts', 'Followers':'followers_count', 'Following':'following_count'}, axis=1, inplace=True)

udf3
```

## Combine All Datasets

```python
# Combine all real users profile datasets
udfs = [udf, udf1, udf2, udf3]
combined_df = pd.concat(udfs, ignore_index=True)

combined_df
```

## Additional Columns

```python
# Create additional columns 
# id
combined_df['id'] = range(1, 1 + len(combined_df)) 

# Bio length
combined_df['bio_length'] = combined_df['bio'].str.len().fillna(-1).astype(int)      # -1 represent if bio unknown

# Number of digit in Username
combined_df['num_digits_username'] = combined_df['username'].str.count(r'\d').fillna(0).astype(int)

# Username_length
combined_df['username_length'] = combined_df['username'].str.len().fillna(-1).astype(int)
# Digit & Uername Ratio
combined_df['digit_ratio_username'] = combined_df['num_digits_username'] / combined_df['username_length'] 
combined_df['digit_ratio_username'] = combined_df['digit_ratio_username'].replace([np.inf, -np.inf], -1).fillna(-1)

# Real accounts
combined_df['fake'] = 0

# Followers & Following ratio
combined_df["followers_count"] = pd.to_numeric(combined_df["followers_count"], errors="coerce").fillna(-1).astype(int)
combined_df["following_count"] = pd.to_numeric(combined_df["following_count"], errors="coerce").fillna(-1).astype(int)
combined_df['followers_following_ratio'] = np.select(
    [
        (combined_df['followers_count'] == -1) | (combined_df['following_count'] == -1),      # Condition unknown
        (combined_df['followers_count'] == 0) | (combined_df['following_count'] == 0)     # Condition 0
    ],
    [
        -1,         # result  if following/followers is unknown, set -1
        0           # result  if following/followers is 0, set 0
    ],
    default=combined_df['followers_count'] / combined_df['following_count']
) 
combined_df['followers_following_ratio'] = combined_df['followers_following_ratio'].round(2)       # round result

# Reorder columns
combined_df = combined_df[['id', 'username', 'name', 'account_created_at', 
                           'bio', 'bio_length', 'num_posts', 'followers_count', 
                           'following_count', 'username_length', 'num_digits_username',
                           'digit_ratio_username', 'followers_following_ratio', 'fake']]

combined_df
```

## Cleaned Datasets

```python
# Get dataset information
print(combined_df.info())
```

```python
# Check if numeric & text for the column
print(combined_df.dtypes)
```

```python
# conver column elements to numberic


combined_df["num_posts"] = pd.to_numeric(combined_df["num_posts"], errors="coerce").fillna(-1).astype(int)

# Convert "created_at" to datetime
combined_df["account_created_at"] = pd.to_datetime(combined_df["account_created_at"], errors="coerce")

print("\nUpdated Data Types:")
print(combined_df.dtypes)
```

```python
# Check for any missing values
print(combined_df.isnull().sum())
```

```python
# Fill missing values
combined_df["name"] = combined_df["name"].fillna("NA")
combined_df["username"] = combined_df["username"].fillna("NA")
combined_df["num_posts"] = combined_df["num_posts"].fillna(-1)

combined_df["account_created_at"] = combined_df["account_created_at"].fillna(-1)
combined_df["bio"] = combined_df["bio"].fillna("NA")

# Verify missing values are handled
print(combined_df.isnull().sum())
```

```python
# Check Duplicate
duplicates = combined_df.duplicated().sum()
print(f"\nOriginal Number of Duplicates:{duplicates}")

# Remove Duplicate
combined_df = combined_df.drop_duplicates()

# Verify duplicates are removed
print(f"Number of of Duplicates Remaining: {combined_df.duplicated().sum()}")
```

```python
combined_df
```

## Convert To CSV

```python
# Save all cleaned datasets
Cleaned_file_path = "user_cleaned_dataset.csv"
combined_df.to_csv(Cleaned_file_path, index=False)

print(f"\nCleaned dataset saved to: {Cleaned_file_path}")
```
